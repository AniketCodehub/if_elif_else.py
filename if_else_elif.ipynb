```python
side1 = int(input("enter the side1 of triangle: "))
side2 = int(input("enter the side2 of a triangle: "))
side3 = int(input("enter the side3 of a triangle: "))
if (side1 == side2 == side3):
    print("The Given triangle is an equilateral triangle")
elif (side1 == side2 or side2 == side3 or side1 == side3):
    print("triangle is isosceles triangle")
elif (side1 != side2 and side2 != side3 and side1 != side3):
    print("triangle is an scalene triangle")
else:
    print("invalid triangle")
```

```python
# calculator building logic
num1 = int(input("enter a number: "))
operator = input("enter the operator: ")
num2 = int(input("enter a number: "))
if (operator == "+"):
    print(num1 + num2)
elif(operator == "-"):
    print(num1 - num2 or num2 - num1)
elif(operator == "*"):
    print(num1 * num2)
elif(operator == "/"):
    print(num1 / num2 or num2 / num1)
```
